##### funcoes executam a simulacao de qualquer protocolo

```python
def p_quadrado(phi):
    zero = np.cos(phi/2)**2
    um = np.sin(phi/2)**2
    return zero, um

def run_qbit(qc):
    job = AerSimulator().run(qc, shots=10000)
    result = job.result()
    counts = result.get_counts()
    
    if len(counts) != 1: 
        zero = counts["0"]
        um = counts["1"]
    else:
        zero , um = p_quadrado(phi)

    return zero, um
```

```python
phis = [0, 3*np.pi/4, np.pi/2, np.pi, 3*np.pi/2]
phis = list(map(lambda x: round(x, 2), phis))

for phi in phis:
    qc = mz_protocol(phi)

    zeros_simulados , ums_simulados= run_qbit(qc)
    zeros_simulados /= 100
    ums_simulados /= 100

    zero, um = p_quadrado(phi)
    zero *= 100
    um   *= 100

    print("-----------------------------------------------------")
    print(" ")
    print(f"angulo phi = {phi:.2f} radianos")
    print(f"Probabilidades simuladas     | 0: {zeros_simulados:.2f}% | 1: {ums_simulados:.2f}%")
    print(f"Probabilidades teóricas      | 0: {zero:.2f}% | 1: {um:.2f}%")
print("")
print("-----------------------------------------------------")
```
